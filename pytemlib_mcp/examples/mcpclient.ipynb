```python
import os

from mcpclient_tems import RemoteMCPClient
from pytemlib_mcp.defaults import HOST, PORT
```

```python
mcp = RemoteMCPClient(HOST, PORT)
print("Connected:", mcp.initialize()["result"]["serverInfo"])
for tool in mcp.list_tools()["result"]["tools"]:
    print(" -", tool["name"])
```

```python
filename = os.path.abspath(os.path.join("..", "..", "example_data", "EDS-STO.emd"))
mcp.call_tool("open_tem_file", {"filename": filename})
```

```python
mcp.call_tool("get_dataset_info", {"filename": filename, "dataset_key": "SuperX"})
```

```python
mcp.close()
```
